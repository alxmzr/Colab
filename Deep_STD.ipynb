## Data Acquisition

### Subtask:
Specify the financial instrument and load historical OHLCV data for analysis.

**Reasoning**:
I will import the necessary libraries and download historical OHLCV data for Apple (AAPL) using yfinance to fulfill the data acquisition subtask.

```python
import pandas as pd
import yfinance as yf
from datetime import datetime

# Define parameters
symbol = 'AAPL'
start_date = '2020-01-01'
end_date = datetime.now().strftime('%Y-%m-%d')

# Download historical data
data = yf.download(symbol, start=start_date, end=end_date, auto_adjust=True)

# Ensure index is datetime
data.index = pd.to_datetime(data.index)

# Display the first few rows and column information
print(f'Data for {symbol} loaded successfully.')
print(data.info())
print(data.head())
```

```python
# Calculate and display summary statistics for the 'Close' column
print("Summary statistics for 'Close' price:")
display(data['Close'].describe())
```

```python
# Check for missing values in the dataset
print("Missing values per column:")
display(data.isnull().sum())
```
